## Baseline Model

```python
# Separate features and target
X = df_clean.drop("Class", axis=1)
y = df_clean["Class"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training size:", X_train.shape)
print("Testing size:", X_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

# Scale the features
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Scaling complete")
```

```python
from sklearn.linear_model import LogisticRegression

# Train the baseline model
model = LogisticRegression(max_iter=1000)
model.fit(X_train_scaled, y_train)

print("Model trained successfully")
```

```python
from sklearn.metrics import classification_report, confusion_matrix

# Make predictions
y_pred = model.predict(X_test_scaled)

# Check model performance
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

### Baseline Result ###

The logistic regression model detected 55 of 95 fraudulent transactions, giving a fraud recall of 58%. Although overall accuracy was very high, the model still missed 40 fraud cases because of the class imbalance.

### Class-Weighted Logistic Regression ###

```python
# Give more importance to the minority class
weighted_model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced"
)

weighted_model.fit(X_train_scaled, y_train)

# Make predictions
y_pred_weighted = weighted_model.predict(X_test_scaled)

print(confusion_matrix(y_test, y_pred_weighted))
print(classification_report(y_test, y_pred_weighted))
```

### Weighted Model Result ###

Class weighting improved fraud recall from 58% to 87%, reducing missed fraud cases from 40 to 12. However, false positives increased significantly, showing a trade-off between detecting more fraud and incorrectly flagging legitimate transactions.

```python
from sklearn.metrics import precision_score, recall_score, f1_score

# Compare both models
results = pd.DataFrame({
    "Model": ["Baseline", "Class Weighted"],
    "Precision": [
        precision_score(y_test, y_pred),
        precision_score(y_test, y_pred_weighted)
    ],
    "Recall": [
        recall_score(y_test, y_pred),
        recall_score(y_test, y_pred_weighted)
    ],
    "F1 Score": [
        f1_score(y_test, y_pred),
        f1_score(y_test, y_pred_weighted)
    ]
})

results
```

```python
from sklearn.ensemble import RandomForestClassifier

# Train a Random Forest model
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

print("Random Forest trained successfully")
```

```python
# Evaluate Random Forest
y_pred_rf = rf_model.predict(X_test)

print(confusion_matrix(y_test, y_pred_rf))
print(classification_report(y_test, y_pred_rf))
```

### Random Forest Result ###

Random Forest detected 69 of 95 fraud cases with 97% precision and 73% recall. It provided a better balance between fraud detection and false positives than the logistic regression models.
